```python
import pandas as pd 
import numpy as np
import logging 

logging.basicConfig(
    level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s',
    filename='quant.log'
)
logger = logging.getLogger(__name__)
```

```python
# loading data 
df = pd.read_csv('cleaned_credit_risk_dataset.csv')
logger.info('Data loaded successfully')
df
```

```python
# select numerical features 
num_cols = ['age', 'income', 'emp_length', 'loan_grade', 'loan_amnt', 'loan_int_rate', 'loan_status', 'loan_percent_income', 'default_on_file', 'cred_hist_length']

# initialize results list 
results = []

# for each column 
for col in num_cols:
    # drop missing values
    data = df[col].dropna()
    
    n = len(data)
    mean = data.mean()
    std = data.std(ddof=1)
    
    # standard error
    se = std / np.sqrt(n)
    
    # 95% confidence interval 
    ci_lower = mean - 1.96 * se
    ci_upper = mean + 1.96 * se
    
    results.append({
        "feature": col,
        "mean": round(mean, 2),
        "std_dev": round(std, 2),
        "standard_error": round(se, 2),
        "ci_lower_95": round(ci_lower, 2),
        "ci_upper_95": round(ci_upper, 2)
    })
    logger.info(f'Results for {col} added to results list.')

ci_df = pd.DataFrame(results)
logger.info('Printing ci results dataframe.')
ci_df
```
